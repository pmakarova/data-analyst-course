[1588. Sum of All Odd Length Subarrays](https://leetcode.com/problems/sum-of-all-odd-length-subarrays/description/)

```python
class Solution(object):
    def sumOddLengthSubarrays(self, arr):
        """
        :type arr: List[int]
        :rtype: int
        """
        p = [0]
        res = 0
        n = len(arr)
        for i in range(n):
            p.append(p[-1] + arr[i])

        for l in range(n):
            r = l
            while r < n:
                res += p[r+1] - p[l]
                r += 2
        return res
```

```python
def sumOddLengthSubarrays(arr):
    """
    :type arr: List[int]
    :rtype: int
    """
    res = 0
    n = len(arr)
    for i in range(n):
        total_subs = (i+1)*(n-i)
        odd_subs = (total_subs + 1) // 2 
        res += odd_subs * arr[i]
    return res
```

```python
arr = [1,4,2,5,3]
print(sumOddLengthSubarrays(arr))
```

[2485. Find the Pivot Integer](https://leetcode.com/problems/find-the-pivot-integer/)

```python
def pivotInteger(n):
    """
    :type n: int
    :rtype: int
    """
    p = [0] * (n + 1)
    for i in range(1, n + 1):
        p[i] = p[i-1] + i
    for i in range(len(p) - 1, int(p[-1]**0.5)-1, -1):
        if p[-1] - p[i] == p[i-1]:
            return i
    return -1
```

```python
n = 4
print(pivotInteger(n))
```

[2389. Longest Subsequence With Limited Sum](https://leetcode.com/problems/longest-subsequence-with-limited-sum/description/)

```python
def answerQueries(nums, queries):
    """
    :type nums: List[int]
    :type queries: List[int]
    :rtype: List[int]
    """
    nums.sort()
    n = len(nums)
    ans = []
    for i in range(1, n):
        nums[i] += nums[i-1]
    for q in queries:
        left = 0
        right = n
        while left < right:
            mid = (left + right)//2
            if nums[mid] <= q:
                left = mid + 1
            else:
                right = mid
        ans.append(left)

    return ans
```

```python
import bisect
def answerQueries(nums, queries):
    nums.sort()
    n = len(nums)
    ans = []
    for i in range(1, n):
        nums[i] += nums[i-1]
    for q in queries:
        ans.append(bisect.bisect_right(nums, q))
    return ans
```

```python
answerQueries([4,5,2,1], [3,10,21])
```

[1310. XOR Queries of a Subarray](https://leetcode.com/problems/xor-queries-of-a-subarray/description/)

```python
def xorQueries(arr, queries):
    p = [0] * (len(arr) + 1)
    for i, x in enumerate(arr):
        p[i + 1] = p[i] ^ x
    return [p[right+1] ^ p[left] for left, right in queries]
```

```python
arr = [4,8,2,10]
queries = [[2,3],[1,3],[0,0],[0,3]]

print(xorQueries(arr, queries))
```

[2680. Maximum OR](https://leetcode.com/problems/maximum-or/description/)

```python
def maximumOr(nums, k):
    n = len(nums)
    p = [0]*(n+1)
    s = [0]*(n+1)
    for i in range(n):
        p[i+1] = p[i] | nums[i]
    for i in range(n - 1, -1, -1):
        s[i] = s[i+1] | nums[i]
    ans = 0
    for i in range(n):
        ans = max(ans, p[i] | nums[i] << k | s[i + 1])
    return ans
```

```python
nums = [8,1,2]
k = 2
print(maximumOr(nums, k))
```

[2381. Shifting Letters II](https://leetcode.com/problems/shifting-letters-ii/description/)

```python
def shiftingLetters(s, shifts):
    n = len(s)
    diff = [0]*(n+1)

    for start, end, direction in shifts:
        val = 1 if direction == 1 else -1
        diff[start] += val
        diff[end+1] -= val

    res = []
    curr_shift = 0

    for i in range(n):
        curr_shift += diff[i]

        orig_code = ord(s[i]) - ord('a')
        new_code = (orig_code + curr_shift) % 26
        res.append(chr(ord('a') + new_code))
    return "".join(res)
```

```python
s = "abc"
shifts = [[0,1,0],[1,2,1],[0,2,1]]
shiftingLetters(s, shifts)
```
