# Бинарный поиск

Бинарный поиск (или двоичный поиск) — это высокоэффективный алгоритм поиска нужного значения в отсортированном массиве данных. Вместо последовательной проверки каждого элемента, он на каждом шаге делит область поиска пополам, отбрасывая ту часть, где искомого значения точно нет.

Ключевое правило: массив для поиска обязательно должен быть отсортирован заранее.
