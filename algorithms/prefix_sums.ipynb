# Префиксные суммы

Алгоритм префиксных сумм используется для быстрого вычисления суммы элементов массива на заданном отрезке $[l, r]$. Он применяется в задачах, где требуется многократно отвечать на запросы о сумме или изменять значения на отрезках массива.

Применение префиксных сумм снижает время ответа на один запрос с $O(n)$ до $O(1)$. 

Если задано $q$ запросов к массиву длины $n$, наивный подход работает за время $O(q \cdot n)$

Алгоритм префиксных сумм требует $O(n)$ на предварительную построение массива и $O(1)$ на каждый запрос, что дает итоговую сложность $O(n + q)$.

---

## Математическое описание алгоритма

Пусть задан исходный массив $a$ длины $n$. Массив префиксных сумм $P$ длины $n+1$ строится по следующему правилу:$$P_0 = 0$$ $$P_i = P_{i-1} + a_{i-1} \quad \text{для } 1 \le i \le n$$

Тогда сумма элементов исходного массива на отрезке $[l, r]$ (где $0 \le l \le r < n$) вычисляется за константное время с помощью формулы:$$S(l, r) = P_{r+1} - P_l$$

---

## Пример



Пусть дан исходный массив $A$ из 5 элементов: $$A = [3, 1, 4, 1, 5]$$

Мы хотим уметь быстро находить сумму на любом отрезке $[l, r]$, например, на отрезке от индекса $l = 1$ до $r = 3$ (то есть сумму элементов $1 + 4 + 1 = 6$).

Шаг 1: Построение массива префиксных сумм

Создадим массив $P$ размером $n + 1$ (в нашем случае $5 + 1 = 6$). Нулевой элемент всегда равен нулю: $P_0 = 0$. Каждый следующий элемент вычисляется по формуле $P_{i+1} = P_i + A_i$.

Вычисляем последовательно:$$P_0 = 0$$ $$P_1 = P_0 + A_0 = 0 + 3 = 3$$ $$P_2 = P_1 + A_1 = 3 + 1 = 4$$ $$P_3 = P_2 + A_2 = 4 + 4 = 8$$ $$P_4 = P_3 + A_3 = 8 + 1 = 9$$ $$P_5 = P_4 + A_4 = 9 + 5 = 14$$

Итоговый массив префиксных сумм:
$$P = [0, 3, 4, 8, 9, 14]$$

Шаг 2: Ответ на запрос

Чтобы найти сумму элементов массива $A$ на отрезке от $l = 1$ до $r = 3$, используем формулу:

$$S(l, r) = P_{r+1} - P_l$$

Подставляем наши индексы:

$$S(1, 3) = P_{3+1} - P_1 = P_4 - P_1$$

Смотрим значения в массиве $P$:

$$P_4 = 9$$
$$P_1 = 3$$

Вычисляем результат:
$$S(1, 3) = 9 - 3 = 6$$

Проверяем вручную по исходному массиву $A$: $A_1 + A_2 + A_3 = 1 + 4 + 1 = 6$. Ответ совпал.



Решение кодом:

```python
A = [3, 1, 4, 1, 5]

n = len(A)
P = [0]*(n+1)

for i in range(n):
    P[i+1] = P[i] + A[i]

print("Исходный массив А: ", A)
print("Массив префексных сумм P: ", P)

l = 1
r = 3

sum_lr = P[r+1] - P[l]
print(f"Сумма на отрезке от индекса {l} до {r} равна: {sum_lr}")
```

## Общий вид алгоритма

```python
def build_prefix_sums(a):
    n = len(a)
    P = [0]*(n+1)

    for i in range(n):
        P[i+1] = P[i] + a[i]
    return P

def get_range_sum(P, l, r): 
    return P[r+1] - P[l]
```

```python
# Пример использования
a = [1, 0, -1, 4]
q = 2

P = build_prefix_sums(a)

print(get_range_sum(P, 0, 2))  # от 0 до 2
print(get_range_sum(P, 1, 3))  # от 1 до 3
```

# Задачи на префексные суммы

### 1. Базовые префиксные и суффиксные суммы

Применяются, когда нужно быстро получать суммы на префиксах или суффиксах массива для сравнения частей массива между собой:
1. Вычисление текущей суммы одномерного массива ([Running Sum of 1d Array](https://leetcode.com/problems/running-sum-of-1d-array/description/)).


```python
class Solution(object):
    def runningSum(self, nums):
        """
        :type nums: List[int]
        :rtype: List[int]
        """
        P = [0]*(len(nums)+1)
        for i in range(len(nums)):
            P[i+1] = P[i] + nums[i]
        return P[1:] 
```

2. Задачи, требующие вычисления как префиксных, так и суффиксных сумм, например, для нахождения разницы между левой и правой суммами ([Left and Right Sum Differences](https://leetcode.com/problems/left-and-right-sum-differences/description/)).



```python
# используя предыдущий аналог
class Solution(object):
    def leftRightDifference(self, nums):
        """
        :type nums: List[int]
        :rtype: List[int]
        """
        P = [0]*(len(nums)*2+1)
        for i in range(len(nums)):
            P[i+1] = P[i] + nums[i]
            P[-i-2] = P[-i-1] + nums[-i-1]
        return [abs(a - b) for a, b in zip(P[:len(nums)], P[-len(nums):])]
```

```python
# один раз посчитать всю сумму
class Solution(object):
    def leftRightDifference(self, nums):
        """
        :type nums: List[int]
        :rtype: List[int]
        """
        left_sum = 0
        right_sum = sum(nums)
        res = []
    
        for num in nums:
            right_sum -= num
            res.append(abs(left_sum - right_sum))
            left_sum += num
        return res
```

3. Поиск специфического индекса (например, где сумма слева равна сумме справа) с помощью префиксных и суффиксных сумм ([Find the Middle Index in Array](https://leetcode.com/problems/find-the-middle-index-in-array/)).


```python
class Solution(object):
    def findMiddleIndex(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        n = len(nums) 
        pref = [0]*(n+1)
        suff = [0]*(n+1)
        for i in range(n):
            pref[i+1] = pref[i] + nums[i]
            suff[n-i-1] = suff[n-i] + nums[n-i-1]
        for i in range(n):
            if pref[i+1] == suff[i]: return i
        return -1
```

4. Задачи на расчет аккумулированных значений с индексацией с единицы, например, вычисление высоты ([Find the Highest Altitude](https://leetcode.com/problems/find-the-highest-altitude/)).

```python
class Solution(object):
    def largestAltitude(self, gain):
        """
        :type gain: List[int]
        :rtype: int
        """
        p = 0
        maxi = 0
        for x in gain:
            p += x
            if maxi < p: maxi = p
        return maxi
```

### 2. Массовые обновления на отрезке

Это очень важная идея для задач, где поступает $q$ запросов вида «прибавить число к элементам на отрезке $[l, r]$».

1. [Points That Intersect With Cars](https://leetcode.com/problems/points-that-intersect-with-cars/)


```python
class Solution(object):
    def numberOfPoints(self, nums):
        """
        :type nums: List[List[int]]
        :rtype: int
        """
        n = len(nums)
        maxi = 0
        count = 0
        for i in range(n):
            if maxi < nums[i][1]: maxi = nums[i][1]
        p = [0]*(maxi+2)
        for q in nums:
              p[q[0]] += 1
              p[q[1]+1] -= 1
        for i in range(1, maxi+2):
              p[i] += p[i-1]
              if p[i] > 0: count += 1
        return count
```

2. [Check if All the Integers in a Range Are Covered](https://leetcode.com/problems/check-if-all-the-integers-in-a-range-are-covered/)

```python
class Solution(object):
    def isCovered(self, ranges, left, right):
        """
        :type ranges: List[List[int]]
        :type left: int
        :type right: int
        :rtype: bool
        """
        p = [0]*52

        for s, e in ranges:
            p[s]+=1
            p[e+1]-=1
        current = 0
        for i in range(1, right+1):
            current+=p[i]
            if i >= left and current==0:
                return False
        return True
```

### 3. Префиксные операции отличные от суммы (например, префиксный XOR)


Алгоритм префиксных сумм работает с любой обратимой ассоциативной операцией. Пишется он точно так же, как и обычная сумма.   

1. Используется для быстрого нахождения $\text{XOR}$ на отрезке и поиска числа $k$, максимизирующего значение, где $k = \text{prefix XOR} \oplus \text{max Number}$ ([Maximum XOR for Each Query](https://leetcode.com/problems/maximum-xor-for-each-query/description/)).

```python
class Solution(object):
    def getMaximumXor(self, nums, maximumBit):
        """
        :type nums: List[int]
        :type maximumBit: int
        :rtype: List[int]
        """
        n = len(nums)
        x = 0
        res = [0]*n
        mask = (1 << maximumBit) - 1
        for i in range(n):   
            x ^= nums[i]   
            res[n-i-1] = x ^ mask
        return res
```

2. Применяется для проверки равенства $\text{XOR}$ на двух смежных отрезках $[i, j-1]$ и $[j, k]$ ([Count Triplets That Can Form Two Arrays of Equal XOR](om/problems/count-triplets-that-can-form-two-arrays-of-equal-xor/)).

```python
def countTriplets(arr):
    """
    :type arr: List[int]
    :rtype: int
    """
    n = len(arr)
    count = 0
    ans = []
    for i in range(n-1):
        a = 0
        for j in range(i+1, n):
            b = 0
            a ^= arr[j-1]
            for k in range(j, n):
                b ^= arr[k]
                if a == b:
                    count += 1
                    ans.append((i,j,k))
    return count
```

```python
def countTriplets(arr):
    """
    :type arr: List[int]
    :rtype: int
    """
    n = len(arr)
    count = 0
    ans = []
    for i in range(n-1):
        xor = arr[i]
        for j in range(i+1, n):
            xor ^= arr[j] 
            if xor == 0:
                count += j-i
    return count
```

### 4. Двумерные префиксные суммы 

Применяются для быстрого нахождения суммы чисел в подматрице (задача 1314. Matrix Block Sum).
1. Заранее считается сумма элементов в подматрице от координат $(0, 0)$ до $(i, j)$ и сохраняется в двумерный массив $p[i][j]$.
2. Сумма произвольной подматрицы находится за $O(1)$ путем комбинирования (добавления и вычитания) ранее посчитанных «краев» подматриц в массиве $p$.

### 5. Оптимизация перебора (в связке с бинарным поиском или двумя указателями)

Префиксные суммы позволяют за $O(1)$ проверять, удовлетворяет ли подотрезок определенным условиям (например, содержит ли он не более $k$ определенных символов). 
1. Наивный поиск таких подотрезков работает долго, с асимптотикой $O(n^2)$.
2. Использование префиксных сумм вместе с бинарным поиском улучшает время до $O(n \log n)$, а в комбинации с методом двух указателей позволяет решить задачу за линейное время $O(n)$ (задача 2024. Maximize the Confusion of an Exam).
